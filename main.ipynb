```python
your_sentence = input("Enter your sentence")
letters=[]
number=[]
for char in your_sentence:
    if char.isalpha():
        letters.append(char)
    elif char.isdigit():
        number.append(int(char))

word= "".join(letters)
total = sum(number)
print(word)
print(total)


```

```python

password = input("Enter your password: ")

has_upper = False
has_special = False

special_chars = "@#$%^&*!?"

for char in password:
    if char.isupper():
        has_upper = True

    if char in special_chars:
        has_special = True

if 8 <= len(password) <= 16 and has_upper and has_special:
    print("Password is Secure.")
else:
    print("Password is Not Secure.")
```

```python

input1 = input("Enter the first list of names: ")
input2 = input("Enter the second list of names: ")

list1 = [name.strip() for name in input1.split(",")]
list2 = [name.strip() for name in input2.split(",")]

common_names = []

for name in list1:
    if name in list2 and name not in common_names:
        common_names.append(name)

print("Common names:", common_names)
```

```python

n = int(input("Enter a multi-digit number: "))

while n >= 10:
    total = 0

    for digit in str(n):
        total += int(digit)

    n = total

print("sum", n)
```

```python

numbers = [1, 1, 2, 3, 3, 3, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

result = []

for num, count in frequency.items():
    result.append((num, count))

print(result)

```

```python

n = int(input("Enter a positive integer: "))

factorial = 1
total = 0

for i in range(1, n + 1):
    factorial *= i
    total += i

print("Factorial is", factorial)
print("Sum is", total)

if total != 0 and factorial % total == 0:
    print(f"{factorial} is divisible by {total}.")
else:
    print(f"{factorial} is not divisible by {total}.")
```

```python

store = {
    "apple": 0.5,
    "bread": 2.0,
    "milk": 1.5
}

cart = []

while True:
    action = input("Enter Add, Remove, or Done: ").lower()

    if action == "done":
        break

    if action == "add":
        item = input("Enter the item name: ").lower()

        if item in store:
            cart.append(item)
            print(item, "added to cart.")
        else:
            print("Error: Item not found in store.")

    elif action == "remove":
        item = input("Enter the item name: ").lower()

        if item in cart:
            cart.remove(item)
            print(item, "removed from cart.")
        else:
            print("Error: Item not found in cart.")

    else:
        print("Invalid action.")

total = 0

for item in cart:
    total += store[item]

print(f"Total Bill: ${total:.2f}")
```

```python

numbers = [1, 5, 2, 6, 3]

peaks = []

for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i - 1] and numbers[i] > numbers[i + 1]:
        peaks.append(numbers[i])

print("Peaks:", peaks)
```

```python

original = {
    "A": 1,
    "B": 2,
    "C": 1
}

flipped = {}

for key, value in original.items():
    if value not in flipped:
        flipped[value] = []

    flipped[value].append(key)

print(flipped)
```

```python

text = input("Enter a string: ").lower()

vowels = "aeiou"
positions = {}

for i in range(len(text)):
    char = text[i]

    if char in vowels:
        if char not in positions:
            positions[char] = []

        positions[char].append(i)

print(positions)
```
